## **Rule Validation**

```python
# ==========================================
# RULE VALIDATION
# ==========================================

# Reset status
df["Status"] = "Valid"
df["Alasan"] = ""

# Fungsi validasi
def validasi(nama_kolom, nama_batas, nama_rule):

    nilai = df[nama_kolom]
    batas = df[nama_batas]

    mask_verifikasi = (
        (nilai > batas) &
        (nilai <= batas * (1 + TOLERANSI))
    )

    mask_tidak_valid = (
        nilai > batas * (1 + TOLERANSI)
    )

    df.loc[
        mask_verifikasi & (df["Status"] == "Valid"),
        "Status"
    ] = "Perlu Verifikasi"

    df.loc[
        mask_verifikasi,
        "Alasan"
    ] += f"{nama_rule} mendekati/melebihi batas; "

    df.loc[
        mask_tidak_valid,
        "Status"
    ] = "Tidak Valid"

    df.loc[
        mask_tidak_valid,
        "Alasan"
    ] += f"{nama_rule} melebihi toleransi; "

# ==========================================
# MENJALANKAN VALIDASI
# ==========================================

validasi(
    "Total Hasil Panen (ton)",
    "Batas Hasil Panen",
    "Hasil Panen"
)

validasi(
    "Total Pupuk Kimia (kg)",
    "Batas Pupuk Kimia",
    "Pupuk Kimia"
)

validasi(
    "Total Pupuk Organik (kg)",
    "Batas Pupuk Organik",
    "Pupuk Organik"
)

validasi(
    "Total Pestisida (kg)",
    "Batas Pestisida (kg)",
    "Pestisida Bubuk"
)

validasi(
    "Total Pestisida (liter)",
    "Batas Pestisida (liter)",
    "Pestisida Cair"
)

print("Rule Validation selesai.")
```

# **NORMALISASI & PENENTUAN K**

## **Menyiapkan Data untuk Clustering**

```python
# ==========================================
# MENYIAPKAN DATA UNTUK CLUSTERING
# ==========================================

# Load ulang dataset hasil Tahap 1 (opsional, jika lanjut di sesi baru)
# df_valid = pd.read_csv("Dataset_Siap_KMeans.csv")

df_valid = df_valid.copy()

print("Jumlah data valid (sebelum cek NaN Luas Lahan) :", len(df_valid))

# Cek apakah ada Luas Lahan yang kosong
print("Missing Luas Lahan (ha) :", df_valid["Luas Lahan (ha)"].isnull().sum())
```

## **Menghapus Data dengan Luas Lahan Kosong**

```python
# ==========================================
# MENGHAPUS DATA TANPA LUAS LAHAN
# ==========================================
# Luas Lahan wajib ada karena menjadi pembagi
# untuk menghitung variabel produktivitas (per hektar)

sebelum = len(df_valid)

df_valid = df_valid.dropna(subset=["Luas Lahan (ha)"]).copy()

sesudah = len(df_valid)

print("Sebelum :", sebelum)
print("Sesudah :", sesudah)
print("Data Terhapus :", sebelum - sesudah)
```

## **Feature Engineering Produktivitas (per Hektar)**

```python
# ==========================================
# FEATURE ENGINEERING PRODUKTIVITAS
# Variabel dibuat "per hektar" agar tidak bias
# terhadap luas lahan (petani lahan luas otomatis
# punya total hasil/pupuk lebih besar meskipun
# produktivitasnya belum tentu lebih baik)
# ==========================================

df_valid["Produktivitas (ton/ha)"] = (
    df_valid["Total Hasil Panen (ton)"] / df_valid["Luas Lahan (ha)"]
)

df_valid["Pupuk Kimia per ha (kg/ha)"] = (
    df_valid["Total Pupuk Kimia (kg)"] / df_valid["Luas Lahan (ha)"]
)

df_valid["Pupuk Organik per ha (kg/ha)"] = (
    df_valid["Total Pupuk Organik (kg)"] / df_valid["Luas Lahan (ha)"]
)

df_valid["Pestisida per ha (kg/ha)"] = (
    df_valid["Total Pestisida (kg)"] / df_valid["Luas Lahan (ha)"]
)

df_valid["Pestisida per ha (liter/ha)"] = (
    df_valid["Total Pestisida (liter)"] / df_valid["Luas Lahan (ha)"]
)

print("Variabel produktivitas per hektar berhasil dibuat.")
```

## **Menentukan Fitur Clustering**

```python
# ==========================================
# FITUR YANG DIGUNAKAN UNTUK K-MEANS
# ==========================================

fitur_cluster = [
    "Produktivitas (ton/ha)",
    "Pupuk Kimia per ha (kg/ha)",
    "Pupuk Organik per ha (kg/ha)",
    "Pestisida per ha (kg/ha)",
    "Pestisida per ha (liter/ha)"
]

X = df_valid[fitur_cluster].copy()

display(X.describe())

print("\nMissing value pada fitur clustering :")
print(X.isnull().sum())
```

## **Normalisasi**

```python
# ==========================================
# NORMALISASI (STANDARDSCALER)
# K-Means berbasis jarak Euclidean, sehingga
# setiap fitur perlu disamakan skalanya
# (mean = 0, std = 1) agar fitur dengan satuan
# besar (mis. kg/ha) tidak mendominasi fitur
# dengan satuan kecil (mis. ton/ha)
# ==========================================

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=fitur_cluster, index=X.index)

display(X_scaled_df.describe())
```

## **Elbow Method**

```python
# ==========================================
# ELBOW METHOD
# Melihat penurunan WCSS (Within-Cluster Sum
# of Squares / inertia) di setiap nilai K untuk
# mencari titik "siku" penurunan yang melandai
# ==========================================

from sklearn.cluster import KMeans

K_range = range(2, 11)

wcss = []

for k in K_range:

    kmeans_temp = KMeans(n_clusters=k, random_state=42, n_init=10)

    kmeans_temp.fit(X_scaled)

    wcss.append(kmeans_temp.inertia_)

plt.figure(figsize=(8,5))

plt.plot(list(K_range), wcss, marker="o")

plt.title("Elbow Method")

plt.xlabel("Jumlah Cluster (K)")

plt.ylabel("WCSS (Inertia)")

plt.xticks(list(K_range))

plt.grid(True)

plt.show()
```

## **Silhouette Score Tiap K**

```python
# ==========================================
# SILHOUETTE SCORE UNTUK SETIAP K
# Mengukur seberapa baik setiap data ditempatkan
# di clusternya sendiri dibanding cluster lain.
# Nilai mendekati 1 = pengelompokan baik,
# mendekati 0 = tumpang tindih antar cluster,
# negatif = kemungkinan salah cluster
# ==========================================

from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in K_range:

    kmeans_temp = KMeans(n_clusters=k, random_state=42, n_init=10)

    labels_temp = kmeans_temp.fit_predict(X_scaled)

    skor = silhouette_score(X_scaled, labels_temp)

    silhouette_scores.append(skor)

    print(f"K = {k} -> Silhouette Score = {skor:.4f}")

plt.figure(figsize=(8,5))

plt.plot(list(K_range), silhouette_scores, marker="o", color="green")

plt.title("Silhouette Score untuk Setiap K")

plt.xlabel("Jumlah Cluster (K)")

plt.ylabel("Silhouette Score")

plt.xticks(list(K_range))

plt.grid(True)

plt.show()
```

## **Menentukan K Terbaik**

```python
# ==========================================
# MENENTUKAN K TERBAIK
# Dipilih otomatis berdasarkan Silhouette Score
# tertinggi. Jika ingin memakai K lain (misalnya
# K=3 untuk interpretasi Rendah/Sedang/Tinggi),
# tinggal ganti nilai K_OPTIMAL secara manual.
# ==========================================

K_OPTIMAL = list(K_range)[int(np.argmax(silhouette_scores))]

print("K terbaik berdasarkan Silhouette Score :", K_OPTIMAL)
print("Silhouette Score pada K terbaik        :", round(max(silhouette_scores), 4))

# Jika ingin override manual, un-comment baris berikut:
K_OPTIMAL = 3
```

# **K-MEANS, EVALUASI, & INTERPRETASI**

## **Membangun Model K-Means Final**

```python
# ==========================================
# MODEL K-MEANS FINAL
# ==========================================

kmeans_final = KMeans(
    n_clusters=K_OPTIMAL,
    random_state=42,
    n_init=10
)

df_valid["Cluster"] = kmeans_final.fit_predict(X_scaled)

print("K-Means selesai dijalankan dengan K =", K_OPTIMAL)

display(df_valid[["Nama"] + fitur_cluster + ["Cluster"]].head(10))
```

## **Evaluasi Model**

```python
# ==========================================
# EVALUASI MODEL
# Catatan: K-Means adalah algoritma unsupervised
# (tidak ada label kelas sebenarnya), sehingga
# tidak dikenal istilah "akurasi" seperti pada
# klasifikasi. Kualitas pengelompokan diukur
# dengan metrik validitas cluster berikut:
#
# 1. Silhouette Score  -> semakin dekat ke 1 semakin baik
# 2. Davies-Bouldin Index -> semakin kecil (dekat 0) semakin baik
# 3. Calinski-Harabasz Index -> semakin besar semakin baik
# ==========================================

from sklearn.metrics import davies_bouldin_score, calinski_harabasz_score

skor_silhouette = silhouette_score(X_scaled, df_valid["Cluster"])

skor_dbi = davies_bouldin_score(X_scaled, df_valid["Cluster"])

skor_ch = calinski_harabasz_score(X_scaled, df_valid["Cluster"])

print("="*55)
print("           EVALUASI HASIL CLUSTERING")
print("="*55)
print(f"Jumlah Cluster (K)          : {K_OPTIMAL}")
print(f"Silhouette Score            : {skor_silhouette:.4f}")
print(f"Davies-Bouldin Index        : {skor_dbi:.4f}")
print(f"Calinski-Harabasz Index     : {skor_ch:.2f}")
print("="*55)
```

```python
# ==========================================
# KARAKTERISTIK RATA-RATA TIAP CLUSTER
# ==========================================

profil_cluster = df_valid.groupby("Cluster")[fitur_cluster].mean().round(2)

display(profil_cluster)
```

## **Pelabelan Cluster Berdasarkan Tingkat Produktivitas**

```python
# ==========================================
# PELABELAN CLUSTER
# Label diberikan otomatis dengan mengurutkan
# rata-rata "Produktivitas (ton/ha)" tiap cluster
# dari yang terendah ke tertinggi
# ==========================================

urutan_cluster = profil_cluster["Produktivitas (ton/ha)"].sort_values().index.tolist()

label_per_k = {
    2: ["Produktivitas Rendah", "Produktivitas Tinggi"],
    3: ["Produktivitas Rendah", "Produktivitas Sedang", "Produktivitas Tinggi"],
    4: ["Produktivitas Rendah", "Produktivitas Sedang", "Produktivitas Tinggi", "Produktivitas Sangat Tinggi"],
    5: ["Produktivitas Sangat Rendah", "Produktivitas Rendah", "Produktivitas Sedang", "Produktivitas Tinggi", "Produktivitas Sangat Tinggi"],
    6: ["Produktivitas Sangat Rendah", "Produktivitas Rendah", "Produktivitas Sedang", "Produktivitas Cukup Tinggi", "Produktivitas Tinggi", "Produktivitas Sangat Tinggi"],
}

if K_OPTIMAL in label_per_k:
    nama_label = label_per_k[K_OPTIMAL]
else:
    nama_label = [f"Kelompok {i+1} (Produktivitas Meningkat)" for i in range(K_OPTIMAL)]

peta_label = {cluster_id: nama_label[i] for i, cluster_id in enumerate(urutan_cluster)}

df_valid["Label Cluster"] = df_valid["Cluster"].map(peta_label)

print("Pemetaan Label Cluster :")
for cluster_id, label in peta_label.items():
    rata2 = profil_cluster.loc[cluster_id, "Produktivitas (ton/ha)"]
    print(f"  Cluster {cluster_id} -> {label}  (rata-rata produktivitas: {rata2:.2f} ton/ha)")

display(df_valid[["Nama", "Produktivitas (ton/ha)", "Cluster", "Label Cluster"]].head(10))
```

## **Visualisasi Cluster (PCA 2D)**

```python
# ==========================================
# VISUALISASI CLUSTER DENGAN PCA
# Mereduksi 5 fitur menjadi 2 dimensi agar
# bisa divisualisasikan dalam scatter plot
# ==========================================

from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=42)

X_pca = pca.fit_transform(X_scaled)

df_valid["PCA1"] = X_pca[:, 0]
df_valid["PCA2"] = X_pca[:, 1]

print(f"Total variansi yang dijelaskan 2 komponen PCA : {pca.explained_variance_ratio_.sum()*100:.2f}%")

plt.figure(figsize=(9,7))

for cluster_id in sorted(df_valid["Cluster"].unique()):
    subset = df_valid[df_valid["Cluster"] == cluster_id]
    plt.scatter(subset["PCA1"], subset["PCA2"], label=peta_label[cluster_id], s=60, alpha=0.8)

pusat_pca = pca.transform(kmeans_final.cluster_centers_)

plt.scatter(pusat_pca[:,0], pusat_pca[:,1], c="black", marker="X", s=250, label="Centroid")

plt.title("Visualisasi Hasil K-Means Clustering (PCA 2D)")

plt.xlabel("Principal Component 1")

plt.ylabel("Principal Component 2")

plt.legend()

plt.grid(True, alpha=0.3)

plt.show()
```
